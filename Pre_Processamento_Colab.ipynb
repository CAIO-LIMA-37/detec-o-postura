```python
# Caminho para o arquivo data.yaml
data_yaml_path = '/content/drive/MyDrive/IC projeto 1 dataset GLR.v3i.yolov11/data.yaml'
```

```python
from ultralytics import YOLO
```

```python
# Carregue o modelo YOLOv8n (nano)
model = YOLO('yolov8n.pt')  # Use 'yolov8s.pt', 'yolov8m.pt', etc., para outros modelos

# Treine o modelo
results = model.train(
    data=data_yaml_path,  # Caminho para o arquivo data.yaml
    epochs=100,           # Número de épocas
    imgsz=640,            # Tamanho da imagem
    batch=16,             # Tamanho do batch
    name='monitoramento_idosos'  # Nome do experimento
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Caminho para o arquivo results.csv
results_csv_path = '/content/runs/detect/monitoramento_idosos/results.csv'

# Carregue os dados do CSV
results = pd.read_csv(results_csv_path)

# Exiba as colunas disponíveis
print(results.columns)
```

```python
plt.rcParams['axes.labelsize'] = 14  # Tamanho dos textos dos eixos
plt.rcParams['axes.titlesize'] = 16  # Tamanho do título
plt.rcParams['xtick.labelsize'] = 12  # Tamanho dos ticks do eixo X
plt.rcParams['ytick.labelsize'] = 12  # Tamanho dos ticks do eixo Y
plt.rcParams['legend.fontsize'] = 12  # Tamanho das legendas
```

### 1. Visualizar as perdas (losses)

Você pode plotar as perdas de treino e validação para ver como elas evoluíram ao longo das épocas. Isso ajuda a identificar overfitting ou underfitting.

```python
# Plotar as funções de perda
plt.figure(figsize=(12, 8))

# Perdas de Treino
plt.plot(results['epoch'], results['train/box_loss'], label='Train Box Loss', linestyle='-', color='blue')
plt.plot(results['epoch'], results['train/cls_loss'], label='Train Cls Loss', linestyle='-', color='green')
plt.plot(results['epoch'], results['train/dfl_loss'], label='Train DFL Loss', linestyle='-', color='orange')

# Perdas de Validação
plt.plot(results['epoch'], results['val/box_loss'], label='Val Box Loss', linestyle='--', color='blue')
plt.plot(results['epoch'], results['val/cls_loss'], label='Val Cls Loss', linestyle='--', color='green')
plt.plot(results['epoch'], results['val/dfl_loss'], label='Val DFL Loss', linestyle='--', color='orange')

# Configurações do gráfico
plt.xlabel('Época')
plt.ylabel('Loss')
plt.title('Função de Perda durante o Treinamento')
plt.legend()
plt.grid()
plt.show()
```

### 2. Visualizar as métricas de precisão e revocação

Você pode plotar a precisão, revocação e mAP para ver como o desempenho do modelo melhorou ao longo do tempo.

```python
plt.figure(figsize=(12, 6))

# Precisão e revocação
plt.plot(results['epoch'], results['metrics/precision(B)'], label='Precisão (Precision)')
plt.plot(results['epoch'], results['metrics/recall(B)'], label='Revocação (Recall)')

# mAP50 e mAP50-95
plt.plot(results['epoch'], results['metrics/mAP50(B)'], label='mAP50')
plt.plot(results['epoch'], results['metrics/mAP50-95(B)'], label='mAP50-95')

plt.xlabel('Época')
plt.ylabel('Métrica')
plt.title('Evolução das Métricas ao Longo das Épocas')
plt.legend()
plt.grid()
plt.show()
```

### 3. Analisar as taxas de aprendizado

Você pode ver como as taxas de aprendizado variaram ao longo do treinamento.

Se você quiser visualizar as três linhas separadamente (mesmo que os valores sejam iguais), pode adicionar pequenos desvios artificiais aos valores de pg1 e pg2 para diferenciá-las no gráfico.

```python
# Adicionar pequenos desvios para diferenciar as linhas
results['lr/pg1'] = results['lr/pg1'] + 0.00001  # Adiciona 0.00001 a pg1
results['lr/pg2'] = results['lr/pg2'] + 0.00002  # Adiciona 0.00002 a pg2

# Plotar o gráfico
plt.figure(figsize=(12, 6))

plt.plot(results['epoch'], results['lr/pg0'], label='Taxa de Aprendizado (pg0)', color='blue', linestyle='-')
plt.plot(results['epoch'], results['lr/pg1'], label='Taxa de Aprendizado (pg1)', color='red', linestyle='--')
plt.plot(results['epoch'], results['lr/pg2'], label='Taxa de Aprendizado (pg2)', color='green', linestyle=':')

plt.xlabel('Época', fontsize=14)
plt.ylabel('Taxa de Aprendizado', fontsize=14)
plt.title('Evolução das Taxas de Aprendizado ao Longo das Épocas', fontsize=16)
plt.legend(fontsize=12)
plt.tick_params(axis='both', labelsize=12)
plt.grid()
plt.show()
```

as colunas lr/pg0, lr/pg1 e lr/pg2 têm exatamente os mesmos valores em todas as épocas. Isso significa que as taxas de aprendizado para os três grupos de parâmetros (pg0, pg1 e pg2) são idênticas ao longo de todo o treinamento.

Por que isso aconteceu?
Configuração do YOLO:

O YOLO pode ter sido configurado para usar a mesma taxa de aprendizado para todos os grupos de parâmetros.

Isso é comum em modelos mais simples ou quando não há necessidade de ajustar taxas de aprendizado diferentes para diferentes partes do modelo.

Arquitetura do modelo:

Se o modelo não tiver grupos de parâmetros com necessidades diferentes (por exemplo, camadas iniciais, intermediárias e finais), o YOLO pode aplicar a mesma taxa de aprendizado para todos.

O que isso significa?
Taxas de aprendizado iguais: Todas as partes do modelo estão sendo treinadas com a mesma taxa de aprendizado.

Comportamento idêntico: Como as taxas são iguais, as curvas de pg0, pg1 e pg2 se sobrepõem completamente no gráfico. É por isso que você só vê uma linha (que na verdade são três linhas sobrepostas).

##Obter as previsões e os rótulos reais

Para gerar a matriz de confusão, precisamos das previsões do modelo (y_pred) e dos rótulos reais (y_true). Vamos usar o método model.predict() para obter as previsões e extrair os rótulos reais das anotações.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Caminho para as imagens de teste
test_images_path = '/content/drive/MyDrive/IC projeto 1 dataset GLR.v3i.yolov11/test/images'

# Obtenha as previsões do modelo
results = model.predict(test_images_path, conf = 0.1)

# Listas para armazenar os rótulos reais e as previsões
y_true = []
y_pred = []
```

```python
# Extraia os rótulos reais e as previsões
for result in results:
    # Rótulos reais (ground truth)
    true_labels = result.boxes.cls.cpu().numpy()  # Classes reais
    y_true.extend(true_labels)

    # Previsões do modelo
    pred_labels = result.boxes.cls.cpu().numpy()  # Classes previstas
    y_pred.extend(pred_labels)

# Converta para arrays numpy
y_true = np.array(y_true)
y_pred = np.array(y_pred)
```

```python
# Calcule a matriz de confusão
cm = confusion_matrix(y_true, y_pred)

# Defina os rótulos das classes em português
classes = ['Deitado', 'Sentado', 'Em pé']  # Substitua pelos nomes das suas classes

# Crie a matriz de confusão personalizada
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)

# Ajuste o tamanho das letras
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlabel('Previsão', fontsize=14)
plt.ylabel('Real', fontsize=14)
plt.title('Matriz de Confusão (Base de Teste)', fontsize=16)

# Exiba a matriz de confusão
plt.show()
```

```python
import os
from PIL import Image

# Caminho para a pasta de teste
test_images_path = '/content/drive/MyDrive/IC projeto 1 dataset GLR.v3i.yolov11/test/images'

# Liste todas as imagens na pasta de teste
image_files = [f for f in os.listdir(test_images_path) if f.endswith(('.jpg', '.jpeg', '.png'))]

# Itere sobre cada imagem e faça inferências
for image_file in image_files:
    # Caminho completo para a imagem
    image_path = os.path.join(test_images_path, image_file)

    # Faça inferências na imagem
    results = model.predict(source=image_path, save=True, conf = 0.1)

    # Exiba a imagem com as detecções
    for result in results:
        result.show()  # Exibe a imagem com as bounding boxes e classes detectadas
```
